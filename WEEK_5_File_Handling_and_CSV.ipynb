```python


file = open("week5_student_scores.csv", "r")

next(file)

total = 0
count = 0

for line in file:
    data = line.strip().split(",")
    math_score = int(data[2])

    total += math_score
    count += 1

average = total / count

print("Average Math Score:", average)

file.close()
```

```python

import pandas as pd

data = pd.read_csv("week5_student_scores.csv")

# create total score
data["total"] = data["math"] + data["english"] + data["science"]

# find highest overall student
highest_student = data.loc[data["total"].idxmax()]

print("Highest Overall Student:", highest_student["name"])
print("Total Score:", highest_student["total"])
```
