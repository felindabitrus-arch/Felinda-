```python


import matplotlib.pyplot as plt
import numpy as np
```

```python


students = ["Felinda", "Chris", "Grace", "David", "Justina", "Favor", "Angel", "John", "Charity", "Joe", "Angel", "Queen", "Racheal", "Sini", " Titus"]
math_scores = [65, 80, 45, 90, 75, 55, 88, 92, 60, 73, 60, 65, 90, 80, 70]
english_scores = [70, 85, 50, 88, 80, 58, 90, 95, 62, 77, 50, 60, 70, 65, 55]
chemistry_scores = [72, 78, 48, 92, 79, 60, 85, 98, 64, 74, 65, 60, 90, 80, 70]
```

```python


plt.figure()
plt.hist(math_scores, bins=5)
plt.title("Distribution of Math Scores")
plt.xlabel("Score Range")
plt.ylabel("Frequency")
plt.show()
```

```python


plt.figure()
plt.scatter(math_scores, english_scores)
plt.title("Math vs English Scores")
plt.xlabel("Math Scores")
plt.ylabel("English Scores")
plt.show()
```

```python


plt.figure()
plt.plot(students, math_scores)
plt.plot(students, english_scores)
plt.title("Math and English Scores Comparison")
plt.xlabel("Students")
plt.ylabel("Scores")
plt.show()
```

```python


plt.figure()

plt.subplot(1, 2, 1)
plt.hist(chemistry_scores, bins=5)
plt.title("chemistry Histogram")

plt.subplot(1, 2, 2)
plt.scatter(math_scores, chemistry_scores)
plt.title("Math vs chemistry")

plt.tight_layout()
plt.show()
```

Histogram
The histogram shows how the students’ scores are distributed across Mathematics, English, and Chemistry. Most scores fall between 60 and 90, showing generally good performance. Only a few students scored below 50. Overall, the class performance is average to high.
Scatter Plot
The scatter plot shows the relationship between scores in different subjects. Students who perform well in one subject often perform well in others. This suggests a positive relationship between the subjects.
Multiple Line Chart
The multiple line chart compares each student’s scores across the three subjects. Some students maintain consistent performance, while others perform better in certain subjects. It helps show trends in students’ results.
Subplot
The subplot displays several charts together in one figure. It helps compare different visualizations of the same data. This makes it easier to understand patterns and overall student performance.
